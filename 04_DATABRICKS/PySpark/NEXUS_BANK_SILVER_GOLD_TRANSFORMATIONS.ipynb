- Inconsistent casing (SUCCESS, success, Success).
- Blank strings.
- Invalid date or amount values.

```python
# STEP 10A - Blank and Whitespace Validation

from pyspark.sql.functions import col, trim, length

for table_name, df in dataframes.items():

    string_columns = [
        field.name
        for field in df.schema.fields
        if field.dataType.simpleString() == "string"
    ]

    print("\n" + "=" * 50)
    print("TABLE:", table_name.upper())

    for column_name in string_columns:

        blank_count = df.filter(
            col(column_name).isNotNull() &
            (length(trim(col(column_name))) == 0)
        ).count()

        if blank_count > 0:
            print(column_name, "-> Blank/Whitespace:", blank_count)
```

##STEP 10D â€“ Validate Date and Amount Formats
Scenario: In a real banking pipeline, malformed dates or amounts can cause failed casts, incorrect calculations, or rejected records.

```python

# STEP 10D - Safe Date and Amount Validation

from pyspark.sql.functions import col, trim, expr

date_columns = {
    "facttransaction": "TransactionDate",
    "factpayment": "PaymentDate",
    "factatmwithdrawal": "WithdrawalDate",
    "factloginactivity": "LoginDate"
}

amount_columns = {
    "facttransaction": "Amount",
    "factpayment": "PaymentAmount",
    "factatmwithdrawal": "WithdrawalAmount"
}

# 1. Validate date values
for table_name, column_name in date_columns.items():

    df = dataframes[table_name]

    invalid_count = df.filter(
        expr(
            f"try_to_timestamp(`{column_name}`, "
            "'yyyy-MM-dd HH:mm:ss.SSSSSSS') IS NULL"
        )
    ).count()

    print(table_name.upper(), "| Invalid dates:", invalid_count)


# 2. Validate amount values
for table_name, column_name in amount_columns.items():

    df = dataframes[table_name]

    invalid_count = df.filter(
        col(column_name).isNull() |
        (trim(col(column_name)) == "") |
        expr(f"try_cast(`{column_name}` AS DECIMAL(18,2)) IS NULL")
    ).count()

    print(table_name.upper(), "| Invalid amounts:", invalid_count)
```

##STEP 10F â€“ Confirm Literal NULL Values

```python
# STEP 10F - Distinguish text "NULL" from actual NULL

from pyspark.sql.functions import col, length, concat, lit

check_columns = {
    "facttransaction": ["TransactionDate", "Amount"],
    "factpayment": ["PaymentDate", "PaymentAmount"],
    "factatmwithdrawal": ["WithdrawalDate", "WithdrawalAmount"],
    "factloginactivity": ["LoginDate"]
}

for table_name, columns in check_columns.items():

    df = dataframes[table_name]

    print("\nTABLE:", table_name.upper())

    for column_name in columns:

        print("\nCOLUMN:", column_name)

        df.filter(
            col(column_name).cast("string") == "NULL"
        ).select(
            concat(lit("["), col(column_name), lit("]")).alias("Value"),
            length(col(column_name)).alias("Length"),
            col(column_name).isNull().alias("Is_Actual_Null")
        ).distinct().show(truncate=False)
```

##STEP 10G â€“ Convert text "NULL" into actual NULL

```python

# STEP 10G - Convert text "NULL" into actual NULL
# Keep original Bronze DataFrames unchanged

from pyspark.sql.functions import col, when, upper, trim

silver_dataframes = {}

for table_name, df in dataframes.items():

    cleaned_df = df

    for column_name, data_type in df.dtypes:

        if data_type == "string":

            cleaned_df = cleaned_df.withColumn(
                column_name,
                when(
                    upper(trim(col(column_name))) == "NULL",
                    None
                ).otherwise(col(column_name))
            )

    silver_dataframes[table_name] = cleaned_df

print("Silver DataFrames Created:", len(silver_dataframes))
```

##STEP 11 â€“ Convert date and amount columns to proper data types
Scenario: Banking transaction dates should be timestamps, and monetary amounts should be decimal values for accurate financial calculations.

```python

# STEP 11 - Convert date and amount columns

from pyspark.sql.functions import col, expr

date_columns = {
    "facttransaction": ["TransactionDate"],
    "factpayment": ["PaymentDate"],
    "factatmwithdrawal": ["WithdrawalDate"],
    "factloginactivity": ["LoginDate"]
}

amount_columns = {
    "facttransaction": ["Amount"],
    "factpayment": ["PaymentAmount"],
    "factatmwithdrawal": ["WithdrawalAmount"]
}

# Convert date columns to TIMESTAMP
for table_name, columns in date_columns.items():

    for column_name in columns:

        silver_dataframes[table_name] = (
            silver_dataframes[table_name].withColumn(
                column_name,
                expr(
                    f"try_to_timestamp(`{column_name}`, "
                    "'yyyy-MM-dd HH:mm:ss.SSSSSSS')"
                )
            )
        )

# Convert amount columns to DECIMAL(18,2)
for table_name, columns in amount_columns.items():

    for column_name in columns:

        silver_dataframes[table_name] = (
            silver_dataframes[table_name].withColumn(
                column_name,
                expr(
                    f"try_cast(`{column_name}` AS DECIMAL(18,2))"
                )
            )
        )

print("Date and Amount Type Conversion Completed")
```

##STEP 13 â€“ Standardize inconsistent text values
Scenario: Banking source systems may store the same business value in different formats, such as debit,  DEBIT, and Debit. We standardize these values in Silver.

```python

# STEP 13 - Standardize selected text columns

from pyspark.sql.functions import col, trim, initcap, when

text_columns = {
    "dimcustomer": [
        "FirstName", "LastName", "Gender",
        "City", "State", "CustomerSegment", "CustomerStatus"
    ],
    "dimbranch": [
        "BranchName", "City", "State", "BranchType", "BranchStatus"
    ],
    "dimaccount": [
        "AccountType", "AccountStatus"
    ],
    "dimcard": [
        "CardType", "CardNetwork", "CardStatus"
    ],
    "dimmerchant": [
        "MerchantCategory", "MerchantStatus"
    ],
    "facttransaction": [
        "TransactionType", "Currency",
        "TransactionStatus", "Channel"
    ],
    "factpayment": [
        "PaymentMethod", "PaymentStatus"
    ],
    "factatmwithdrawal": [
        "WithdrawalStatus"
    ],
    "factloginactivity": [
        "LoginChannel", "DeviceType", "LoginStatus"
    ]
}

for table_name, columns in text_columns.items():

    df = silver_dataframes[table_name]

    for column_name in columns:

        df = df.withColumn(
            column_name,
            when(
                col(column_name).isNotNull(),
                initcap(trim(col(column_name)))
            ).otherwise(None)
        )

    silver_dataframes[table_name] = df

print("Text Standardization Completed")
```

##STEP 14 â€“ Apply business-standard mappings
First, let's standardize the most important transaction fields using explicit mappings.

```python

# STEP 14 - Standardize transaction business values

from pyspark.sql.functions import col, when, upper

df = silver_dataframes["facttransaction"]

# Transaction Type
df = df.withColumn(
    "TransactionType",
    when(upper(col("TransactionType")) == "DEBIT", "Debit")
    .when(upper(col("TransactionType")) == "CREDIT", "Credit")
    .when(upper(col("TransactionType")) == "WITHDRAWAL", "Withdrawal")
    .when(upper(col("TransactionType")) == "DEPOSIT", "Deposit")
    .when(upper(col("TransactionType")) == "TRANSFER", "Transfer")
    .when(upper(col("TransactionType")) == "PAYMENT", "Payment")
    .when(upper(col("TransactionType")) == "REFUND", "Refund")
    .otherwise(col("TransactionType"))
)

# Transaction Status
df = df.withColumn(
    "TransactionStatus",
    when(upper(col("TransactionStatus")) == "SUCCESSFUL", "Successful")
    .when(upper(col("TransactionStatus")) == "PENDING", "Pending")
    .when(upper(col("TransactionStatus")) == "REVERSED", "Reversed")
    .when(upper(col("TransactionStatus")) == "FAILED", "Failed")
    .when(upper(col("TransactionStatus")) == "SUCCESS", "Successful")
    .otherwise(col("TransactionStatus"))
)

# Channel
df = df.withColumn(
    "Channel",
    when(upper(col("Channel")) == "ATM", "ATM")
    .when(upper(col("Channel")) == "UPI", "UPI")
    .when(upper(col("Channel")) == "POS", "POS")
    .when(upper(col("Channel")) == "ONLINE", "Online")
    .when(upper(col("Channel")) == "NET BANKING", "Net Banking")
    .when(upper(col("Channel")) == "MOBILE BANKING", "Mobile Banking")
    .when(upper(col("Channel")) == "INTERNET", "Internet")
    .when(upper(col("Channel")) == "MOBILE", "Mobile")
    .otherwise(col("Channel"))
)

silver_dataframes["facttransaction"] = df

print("Transaction Business Mapping Completed")
```

##STEP 33 â€“ Foreign Key / Referential Integrity Validation
Scenario: Every Account must reference an existing Customer and Branch. Every Transaction must reference an existing Account.

```python

# STEP 33 - Foreign Key Validation

from pyspark.sql.functions import col

foreign_keys = [
    ("dimaccount", "CustomerID", "dimcustomer", "CustomerID"),
    ("dimaccount", "BranchID", "dimbranch", "BranchID"),
    ("dimcard", "AccountID", "dimaccount", "AccountID"),
    ("facttransaction", "AccountID", "dimaccount", "AccountID"),
    ("facttransaction", "MerchantID", "dimmerchant", "MerchantID"),
    ("factpayment", "AccountID", "dimaccount", "AccountID"),
    ("factatmwithdrawal", "AccountID", "dimaccount", "AccountID"),
    ("factloginactivity", "CustomerID", "dimcustomer", "CustomerID")
]

for child_table, child_key, parent_table, parent_key in foreign_keys:

    child_df = silver_dataframes[child_table]
    parent_df = silver_dataframes[parent_table]

    child_df = child_df.withColumn(
        child_key,
        col(child_key).cast("int")
    )

    parent_keys = parent_df.select(
        col(parent_key).alias("parent_key")
    ).distinct()

    invalid_count = (
        child_df
        .filter(col(child_key).isNotNull())
        .join(
            parent_keys,
            child_df[child_key] == parent_keys["parent_key"],
            "left_anti"
        )
        .count()
    )

    print(
        child_table, "->", parent_table,
        "| Invalid References:", invalid_count
    )
```

##STEP 40 â€“ Create Gold Customer Transaction Summary
Business scenario: A bank wants to understand customer transaction behavior, including total transactions, successful transactions and total transaction amount.

```python

# STEP 40 - Gold Customer Transaction Summary

from pyspark.sql.functions import (
    col, count, sum, when, round
)

transactions = spark.table(
    "workspace.silver.facttransaction"
)

accounts = spark.table(
    "workspace.silver.dimaccount"
)

customers = spark.table(
    "workspace.silver.dimcustomer"
)

gold_customer_summary = (
    transactions
    .join(accounts, "AccountID", "inner")
    .join(customers, "CustomerID", "inner")
    .groupBy(
        customers["CustomerID"],
        customers["FirstName"],
        customers["LastName"],
        customers["CustomerSegment"]
    )
    .agg(
        count("TransactionID").alias("TotalTransactions"),
        sum(
            when(
                col("TransactionStatus") == "Successful", 1
            ).otherwise(0)
        ).alias("SuccessfulTransactions"),
        round(
            sum("Amount"), 2
        ).alias("TotalTransactionAmount")
    )
)

print("Gold Customer Summary Created")
print("Total Customers:", gold_customer_summary.count())

gold_customer_summary.show(10, truncate=False)
```

```python
from pyspark.sql import functions as F

base = "/Volumes/workspace/default/nexus_bank_transfer/"

customer = spark.read.format("delta").load(base + "dimcustomer")
account = spark.read.format("delta").load(base + "dimaccount")
branch = spark.read.format("delta").load(base + "dimbranch")
date = spark.read.format("delta").load(base + "dimdate")
merchant = spark.read.format("delta").load(base + "dimmerchant")

transaction = spark.read.format("delta").load(base + "facttransaction")
payment = spark.read.format("delta").load(base + "factpayment")
atm = spark.read.format("delta").load(base + "factatmwithdrawal")
login = spark.read.format("delta").load(base + "factloginactivity")

checks = [
    ("Customer PK duplicates", customer, "CustomerID"),
    ("Account PK duplicates", account, "AccountID"),
    ("Branch PK duplicates", branch, "BranchID"),
    ("Date PK duplicates", date, "DateKey"),
    ("Merchant PK duplicates", merchant, "MerchantID")
]

for name, df, key in checks:
    duplicates = (
        df.groupBy(key).count()
        .filter(F.col("count") > 1).count()
    )
    print(name, ":", duplicates)

print("\nFOREIGN KEY VALIDATION")

print(
    "Account -> Customer missing:",
    account.join(customer, "CustomerID", "left_anti").count()
)

print(
    "Account -> Branch missing:",
    account.join(branch, "BranchID", "left_anti").count()
)

print(
    "Transaction -> Account missing:",
    transaction.join(account, "AccountID", "left_anti").count()
)

print(
    "Payment -> Account missing:",
    payment.join(account, "AccountID", "left_anti").count()
)

print(
    "ATM -> Account missing:",
    atm.join(account, "AccountID", "left_anti").count()
)

print(
    "Login -> Customer missing:",
    login.join(customer, "CustomerID", "left_anti").count()
)
```

```python
from pyspark.sql import functions as F

base = "/Volumes/workspace/default/nexus_bank_transfer/"
out = base + "star_schema/"

# 1. Read dimensions
customer = spark.read.format("delta").load(base + "dimcustomer")
account  = spark.read.format("delta").load(base + "dimaccount")
branch   = spark.read.format("delta").load(base + "dimbranch")
merchant = spark.read.format("delta").load(base + "dimmerchant")
date_dim = spark.read.format("delta").load(base + "dimdate")

# 2. Add Unknown Date row
unknown_date = spark.createDataFrame(
    [(0, None, None, None, None, None, None, None)],
    date_dim.schema
)

date_dim = date_dim.unionByName(unknown_date)

# 3. Function to add DateKey
def add_date_key(df, date_column):
    return df.withColumn(
        "DateKey",
        F.when(
            F.col(date_column).isNull(), F.lit(0)
        ).otherwise(
            F.date_format(
                F.to_date(F.col(date_column)), "yyyyMMdd"
            ).cast("int")
        )
    )

# 4. Build FactTransaction
transaction = spark.read.format("delta").load(base + "facttransaction")
fact_transaction = (
    add_date_key(transaction, "TransactionDate")
    .join(
        account.select("AccountID", "CustomerID", "BranchID"),
        "AccountID", "left"
    )
)

# 5. Build FactPayment
payment = spark.read.format("delta").load(base + "factpayment")
fact_payment = (
    add_date_key(payment, "PaymentDate")
    .join(
        account.select("AccountID", "CustomerID", "BranchID"),
        "AccountID", "left"
    )
)

# 6. Build FactATMWithdrawal
atm = spark.read.format("delta").load(base + "factatmwithdrawal")
fact_atm = (
    add_date_key(atm, "WithdrawalDate")
    .join(
        account.select("AccountID", "CustomerID", "BranchID"),
        "AccountID", "left"
    )
)

# 7. Build FactLoginActivity
login = spark.read.format("delta").load(base + "factloginactivity")
fact_login = add_date_key(login, "LoginDate")

# 8. Save all dimensions and facts
tables = {
    "DimCustomer": customer,
    "DimAccount": account,
    "DimBranch": branch,
    "DimMerchant": merchant,
    "DimDate": date_dim,
    "FactTransaction": fact_transaction,
    "FactPayment": fact_payment,
    "FactATMWithdrawal": fact_atm,
    "FactLoginActivity": fact_login
}

for name, df in tables.items():
    df.write.format("delta").mode("overwrite").save(out + name)
    print(name, ":", df.count(), "rows saved")

print("STAR SCHEMA BUILD COMPLETED")
```

```python
import os

source = "/Volumes/workspace/default/nexus_bank_transfer/star_schema/"
destination = "gold/star_schema"

tables = [
    "DimCustomer",
    "DimAccount",
    "DimBranch",
    "DimMerchant",
    "DimDate",
    "FactTransaction",
    "FactPayment",
    "FactATMWithdrawal",
    "FactLoginActivity"
]

for table in tables:
    local_folder = source + table
    uploaded = 0

    for root, dirs, files in os.walk(local_folder):
        for filename in files:
            if filename.endswith(".crc"):
                continue

            local_file = os.path.join(root, filename)
            relative_path = os.path.relpath(
                local_file, local_folder
            ).replace("\\", "/")

            adls_path = f"{destination}/{table}/{relative_path}"

            with open(local_file, "rb") as data:
                file_system.get_file_client(
                    adls_path
                ).upload_data(data, overwrite=True)

            uploaded += 1

    print(table, "uploaded files:", uploaded)

print("ALL 9 TABLES UPLOADED TO ADLS GOLD")
```
